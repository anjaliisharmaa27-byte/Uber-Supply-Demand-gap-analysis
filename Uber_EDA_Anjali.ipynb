# **Project Summary -**

This project is based on actual ride request data that was gathered from Uber over a five-day period. Request ID, Pickup Point, Driver ID, Status, Request Timestamp, and Drop Timestamp are the six columns that make up the dataset's 6745 ride entries. Finding and analyzing the supply-demand imbalance in Uber's operations is the primary objective of this project.
The disparity between the number of ride requests received and the number of rides completed is known as the supply-demand gap. Every ride request should, in a perfect world, lead to a successful trip. However, in practice, a lot of requests are not fulfilled because either the driver cancels the trip or there are no cars available at that particular moment. This project specifically looks into this.

Just 2831 trips, or 42% of the total demand, were successfully accomplished out of 6745 ride requests. This indicates that 3914 requests, or over 58%, were not completed. Uber will lose money as a result of this enormous disparity, and customers will have a bad experience.I utilized Excel, SQL, and Python as needed to carry out my analysis. I used Excel to clean up the data, make pivot tables, and develop a dashboard with charts that displayed the difference by time slot, pickup location, and status. To get useful information regarding status distribution, location-wise gaps, time-slot-wise gaps, and root cause analysis, I ran seven SQL queries on the entire dataset. Using Pandas, Matplotlib, and Seaborn, I conducted in-depth exploratory data analysis in Python, producing 20 charts that covered univariate, bivariate, and multivariate analysis.

My research revealed that the supply-demand mismatch is focused in particular times and places rather than being consistent. With 2779 out of 3914 unsuccessful rides, the Early Morning time slot between 5 and 9 AM had the most unmet requests. This is because drivers frequently cancel rides in the early morning, either as a result of low motivation or lengthy travels to the airport. Because there aren't many cars available late at night, the nighttime slot likewise exhibits a large disparity.Airport pickup locations are more problematic than city pickups in terms of location. The primary problem at the airport is that there were 1713 requests with no cars available. When flights arrive early in the morning and people urgently require taxis, this is very challenging. The primary issue at City pickup locations is driver cancellations; drivers canceled 1066 City requests.

Airport pickup in the early morning with the status "No Cars Available" is the worst combination found in this investigation, accounting for 1031 unfulfilled requests alone. Uber's supply-demand issue is essentially represented by this one combination.
A dedicated airport fleet, the introduction of night shift drivers, driver incentives for early morning shifts, and penalties for repeated cancellations during peak hours are some of the measures I have proposed in light of these findings.
I gained practical expertise with data cleaning, pivot table development, SQL querying, and Python-based exploratory data analysis with this project. Additionally, it made it clearer to me how data-driven insights can directly address actual business issues.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# STEP 1: Converting timestamp columns to proper datetime format
ride_data['Request timestamp'] = pd.to_datetime(
    ride_data['Request timestamp'],
    dayfirst=True,
    errors='coerce'
)

ride_data['Drop timestamp'] = pd.to_datetime(
    ride_data['Drop timestamp'],
    dayfirst=True,
    errors='coerce'
)

# STEP 2: Extracting hour from request timestamp
ride_data['Hour'] = ride_data['Request timestamp'].dt.hour

# STEP 3: Extracting day of week
ride_data['Day_of_Week'] = ride_data['Request timestamp'].dt.day_name()

# STEP 4: Creating time slot category from hour
def assign_time_slot(hour):
    if 0 <= hour <= 4:
        return 'Night'
    elif 5 <= hour <= 8:
        return 'Early Morning'
    elif 9 <= hour <= 11:
        return 'Morning'
    elif 12 <= hour <= 16:
        return 'Afternoon'
    elif 17 <= hour <= 20:
        return 'Evening'
    else:
        return 'Late Night'

ride_data['Time_of_Day'] = ride_data['Hour'].apply(assign_time_slot)

# STEP 5: Creating unfulfilled flag
# 1 = ride failed, 0 = ride completed
ride_data['Is_Unfulfilled'] = ride_data['Status'].apply(
    lambda x: 1 if x != 'Trip Completed' else 0
)

print("Data wrangling completed!")
print("New columns added:", ['Hour', 'Day_of_Week', 'Time_of_Day', 'Is_Unfulfilled'])
print(ride_data.head())
````

### What all manipulations have you done and insights you found?

I converted the timestamp columns to proper datetime format. then I extracted the hour and day of week from the request timestamp.I created a Time of day column to categorize rides into night,early morning,morning,afternoon,evening and late night slots. Finally I created an Is_unfulfilled flag to identify rides that were not completed.

````python
# Chart 20: Completion rate by Hour and Pickup Point
hourly_completion = ride_data.groupby(
    ['Hour', 'Pickup point']
).apply(
    lambda x: (x['Status'] == 'Trip Completed').sum() / len(x) * 100
).unstack()

plt.figure(figsize=(12, 5))
plt.plot(hourly_completion.index, hourly_completion['Airport'],
         marker='o', label='Airport', color='#3498db')
plt.plot(hourly_completion.index, hourly_completion['City'],
         marker='s', label='City', color='#e74c3c')

plt.axhline(y=50, color='black', linestyle='--', label='50% benchmark')
plt.title('Completion Rate by Hour and Pickup Point', fontsize=14)
plt.xlabel('Hour of Day')
plt.ylabel('Completion Rate (%)')
plt.xticks(range(0, 24))
plt.legend(title='Location')
plt.tight_layout()
plt.show()
````
